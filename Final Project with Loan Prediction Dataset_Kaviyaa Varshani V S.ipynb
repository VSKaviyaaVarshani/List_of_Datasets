## Column Description:

Please find the below columns names, their description and use cases

1. **Customer_ID** \
*   Description: \
A unique identifier for each customer in the dataset. The ID is randomly generated for each entry and follows the format LPxxxxxx where xxxxxx is a random number between 100000 and 999999.
*  Use Case: \
The Customer_ID allows tracking of individual applicants and is typically used for reference purposes, ensuring each record is distinct.
--------------------
2. **Gender**
*   Description: \
Indicates the gender of the loan applicant, which can either be 'Male' or 'Female'.
*   Use Case: \
This feature can be used to analyze gender-based loan approval trends or gender disparities in financial services.
--------------------
3. **Marital_Status**
*   Description:\
A categorical variable that indicates whether the applicant is married or not. Possible values are 'Yes' (married) and 'No' (unmarried).
*   Use Case: \
Marital status might correlate with loan approval rates, income levels, and financial stability. It can be used to investigate its impact on loan approval decisions.
--------------------
4. **Number_of_Dependents**
*   Description: \
This feature represents the number of dependents that the applicant supports. It can take values 0, 1, 2, or '3+' (indicating more than 2 dependents).
*   Use Case: \
The number of dependents is an important financial indicator and can help determine an applicant's ability to repay a loan. It could also correlate with financial risk assessments.
--------------------
5. **Education**
*   Description: \
Indicates whether the applicant is a graduate or not. Possible values are 'Graduate' and 'Not Graduate'.
*   Use Case: \
Education level can have a strong relationship with income levels and loan repayment abilities. Analyzing education data can help uncover patterns related to loan approval success.
--------------------
6. **Loan_Status**
*   Description: \
This feature tells whether the applicant has applied for the loan or not. It can take the values 'Yes' or 'No'.
*   Use Case: \
The Loan_Status can be used to filter the applicants who have actively applied for a loan and allow analysis of the approval process or qualification criteria.
--------------------
7. **Annual_Income**
*   Description: \
Represents the annual income of the applicant, given in numeric values (e.g., between 1500 and 12000). It has a 10% chance of missing data (NaN).
*   Use Case: \
Income is a key factor in loan approval decisions. It's typically used to assess an applicant's financial capacity to repay the loan. Missing income data represents real-world gaps that need to be handled in data cleaning.
--------------------
8. **Credit_Score**
*   Description: \
A numeric score (between 0 and 5000) that represents the applicant's creditworthiness. It has a 10% chance of being missing (NaN).
*   Use Case: \
The credit score is one of the primary criteria in loan approval processes, used by financial institutions to evaluate the risk of lending to an individual. Missing credit scores indicate potential gaps in the applicant's financial history.
--------------------
9. **Loan_Amount**
*   Description: \
This column contains the amount of money requested by the applicant for the loan (numeric). It has a 10% chance of missing data (NaN).
*   Use Case: \
The loan amount is another critical factor in the approval process, affecting the size and type of the loan. Missing values in loan amount data could indicate incomplete loan requests or issues with the application.
--------------------
10. **Term**
*   Description: \
The term of the loan, which is typically either 180 months or 360 months (representing 15 or 30 years respectively).
*   Use Case: \
The term affects the repayment schedule, and different terms may be offered based on the applicant's financial stability. The term data can help in loan repayment analysis or in determining which terms are more common in various demographic groups.
--------------------
11. **Property_Area**
*   Description: \
A categorical variable representing the area in which the applicant resides. Possible values are 'Urban', 'Semiurban', and 'Rural'.
*   Use Case: \
Property area can be used to analyze how loan approval rates and financial stability vary by region. It also helps identify geographical patterns in loan demand and repayment trends.
--------------------
12. **Loan_Approval**
*   Description: \
Indicates whether the loan was approved or not, with possible values 'Yes' (approved) and 'No' (not approved).
*   Use Case: \
This is the target variable that represents whether the applicant received the loan they applied for. It is the primary outcome in loan prediction models.

```python
def detect_outliers(df_copy, column):
    Q1 = df_copy[column].quantile(0.25)
    Q3 = df_copy[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df_copy[(df_copy[column] < lower_bound) | (df_copy[column] > upper_bound)]
    return outliers

for col in numerical_cols:
    outlier_df = detect_outliers(df_copy, col)
    print(f"Outliers in {col}:")
    print(outlier_df[[col]])
    print()
```

```python
def remove_outliers(df_copy, columns):
    for col in columns:
        Q1 = df_copy[col].quantile(0.25)
        Q3 = df_copy[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        df_copy = df_copy[(df_copy[col] >= lower_bound) & (df_copy[col] <= upper_bound)]
    return df_copy

df_no_outliers = remove_outliers(df_copy, numerical_cols)
print(f"Before removing outliers: {df_copy.shape}")
print(f"After removing outliers: {df_no_outliers.shape}")
```

We can remove outliers using methods such as IQR or z-score or capping.

Outlier handling is not required for this dataset because the value differences are natural. As a precaution, we can create another DataFrame with outliers removed from respective columns, which can be used later if the model's performance is not satisfactory.

If we plan to perform Feature Scaling in upcoming steps, There is no need of treating outliers & skewness.

```python
x_selected = df[['Loan_Status', 'Loan_Amount', 'Credit_Score', 'Term',
       'Annual_Income', 'Number_of_Dependents', 'Property_Area']]
```

## Split data into training and testing

```python
# Split data into training and testing
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x_selected, y, test_size=0.20, random_state=85)
```

## Feature Scaling

```python
# Feature Scaling:
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

## Classifier Model Building

```python
# Import the required libraries and models used for Classification tasks:
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from xgboost import XGBClassifier

# Dictionary of classifiers
classifiers = {
    'Logistic Regression': LogisticRegression(),
    'K-Nearest Neighbors': KNeighborsClassifier(),
    'Decision Tree': DecisionTreeClassifier(),
    'Random Forest': RandomForestClassifier(),
    'Support Vector Classifier': SVC(),
    'Naive Bayes': GaussianNB(),
    'XGBoost': XGBClassifier()
}
```

```python
from sklearn.metrics import accuracy_score, f1_score

# Initialize an empty dictionary to store the results
results = {}

# Iterate through classifiers
for name, model in classifiers.items():
    model.fit(x_train_scaled, y_train)
    y_pred = model.predict(x_test_scaled)
    # Calculate metrics
    accuracy = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    # Store the results
    results[name] = {'Accuracy': accuracy, 'F1 Score': f1}

# Convert the results dictionary to a DataFrame
results_df = pd.DataFrame(results).T
results_df
```

***Naive Bayes Classifier*** is the best performer overall with the highest accuracy (0.5130) and a competitive F1 score (0.5750).

## K-Fold Cross Validation


In K-Fold Cross Validation we split the dataset into k number of subsets known as folds then we perform training on the all the subsets but leave one (k-1) subset for the evaluation of the trained model. In this method, we iterate k times with a different subset reserved for testing purpose each time.

```python
import warnings

warnings.filterwarnings("ignore")
```

```python
from sklearn.model_selection import KFold, cross_val_score

# K-fold cross-validation
k = 5  # Number of folds
kf = KFold(n_splits=k, shuffle=True, random_state=85)

# Print header
print(f"{'Classifier':<25} {'Mean Accuracy':<20} {'Mean F1 Score':<20}")

# Iterate through classifiers and perform cross-validation
for name, clf in classifiers.items():
    accuracy_scores = cross_val_score(clf, x_selected, y, cv=kf, scoring='accuracy')
    f1_scores = cross_val_score(clf, x_selected, y, cv=kf, scoring='f1_weighted')
    print(f"{name:<25} {accuracy_scores.mean():<20.4f} {f1_scores.mean():<20.4f}")
```

**Logistic Regression** is the best performer overall with the highest Mean Accuracy(0.5088) & and a competitive Mean F1 Score (0.4965).

***Hyperparameter Tuning Methods***
---------------------------------

**GridSearchCV** \
GridSearchCV is a systematic approach to hyperparameter tuning. It exhaustively considers all parameter combinations specified in a grid. For example, if you are tuning a Random Forest classifier with two hyperparameters—n_estimators and max_depth—you could define a grid as follows:

n_estimators: [50, 100, 200]
max_depth: [None, 10, 20]
GridSearchCV will train and evaluate the model for every combination of these values, providing the best-performing set of hyperparameters based on cross-validated performance metrics.


**RandomizedSearchCV** \
RandomizedSearchCV, in contrast, randomly samples from the hyperparameter space for a specified number of iterations. This can be more efficient than GridSearchCV, especially when the hyperparameter space is large. Using the same Random Forest example, you might define:

n_estimators: [50, 100, 200]
max_depth: [None, 10, 20]
With RandomizedSearchCV, instead of evaluating every combination, it will randomly choose a subset of these combinations, allowing you to find a good set of hyperparameters with less computational effort.

### Hyperparameter Tuning Using Grid Search

Grid Search exhaustively searches through a specified subset of hyperparameters.

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier

# Example model
model = RandomForestClassifier()

# Hyperparameters to tune
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
}

# Initialize GridSearchCV
grid_search = GridSearchCV(estimator=model, param_grid=param_grid,
                           scoring='f1_weighted', cv=5, n_jobs=-1, verbose=2)

# Fit GridSearchCV
grid_search.fit(x_train_scaled, y_train)

# Best parameters and score
print("Best parameters:", grid_search.best_params_)
print("Best F1 Score:", grid_search.best_score_)
```

```python
# Evaluate the best model on the test set
best_model = grid_search.best_estimator_

y_pred = best_model.predict(x_test_scaled)
```
